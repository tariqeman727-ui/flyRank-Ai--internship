## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human Review

Before applying any recommendation, a content reviewer should verify:
- The page content is still accurate.
- The suggested update matches user intent.
- No important information will be removed.
- SEO changes are appropriate for the page.

### No-Go List

The following actions should NOT be automated:
- Publishing content without human approval.
- Deleting pages automatically.
- Changing important facts or statistics without verification.
- Rewriting sensitive or legal content automatically.

```python
print("Human Review Checklist")
print("- Verify content accuracy")
print("- Check user intent")
print("- Review SEO changes")
print("- Approve before publishing")

print("\nNo-Go List")
print("- No automatic publishing")
print("- No automatic page deletion")
print("- No automatic changes to legal or sensitive content")
```
